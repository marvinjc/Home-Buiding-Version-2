```python
import pandas as pd
url = "https://raw.githubusercontent.com/marvinjc/metricas_de_desempeno/main/Qualitative_Bankruptcy.data.txt"
```

```python
columnas = [
    "riesgo_industrial",
    "riesgo_administrativo",
    "flexibilidad_financiera",
    "credibilidad",
    "competitividad",
    "riesgo_operativo",
    "clase"
]
```

```python
df = pd.read_csv(url, header=None, names=columnas)
```

```python
display(df.head())
```

```python
conteo = df["clase"].value_counts()
porcentaje = df["clase"].value_counts(normalize=True) * 100
```

```python
print(conteo)
print(porcentaje.round(2))
```
